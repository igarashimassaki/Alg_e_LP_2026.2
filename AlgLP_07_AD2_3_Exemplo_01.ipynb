### **JSON (JavaScript Object Notation) é o formato mais popular do mundo para enviar e organizar informações na internet. Pense nele como a versão digital de uma ficha cadastral preenchida: ele serve para que computadores diferentes (como o seu celular e o servidor de um aplicativo) conversem e entendam os mesmos dados de forma rápida.**
### **A grande vantagem do JSON é que ele foi feito para ser lido facilmente tanto por máquinas quanto por seres humanos.**

### **Para entender na prática, imagine que você está preenchendo os dados de um livro em um sistema.**
### **Em vez de escrever um texto corrido, o JSON organiza as informações em pares de "pergunta e resposta" (que na programação chamamos de chave e valor), sempre usando aspas e separados por dois-pontos.**
### **Veja como fica um exemplo de JSON:**

</div>

<div align="justify">

### **Em lingugem de programação, o conceito de chave-valor (key-value) funciona como uma etiqueta presa a um objeto. A chave é o identificador único (a etiqueta) e o valor é o dado guardado (o objeto).No exemplo a seguir, além do nome, elementos comuns como id e email também funcionam perfeitamente como chaves, pois servem para identificar informações específicas sobre o usuário.**

# **EXEMPLO DE CÓDIGO 01:**
# Criando um par chave-valor para um usuário
</div>


```python
usuario = {
    "id": 1025,
    "nome": "Maria",
    "email": "maria@email.com"
}
# Acessando o valor através da chave
print(usuario["nome"])
# Saída: Maria
```

<div align="justify">
### **Estrutura Completa de Chaves e Valores**
### **Olhando novamente para o nosso código, cada uma das chaves cumpre um papel de identificação para o seu respectivo valor:**
• Chave **id:** É o identificador numérico único do usuário.

	• Seu Valor (1025): Um número inteiro. Ele funciona como o "RG" ou "CPF" no sistema, garantindo que este usuário nunca seja confundido com outro, mesmo que tenham o mesmo nome.

• Chave **nome:** É a etiqueta para o nome da pessoa.

	• Seu Valor ("Maria"): Um texto (string) com o nome próprio do usuário.


• Chave **email:** É a etiqueta para o endereço de contato eletrônico.

	• Seu Valor ("maria@email.com"): Um texto (string) contendo o e-mail. Na prática, o e-mail também costuma ser único no sistema, sendo ótimo para buscas rápidas.
  

### **Representação Visual do Armazenamento**

### **Por trás dos panos, o computador organiza esses dados em uma tabela de associação direta:**

|Chave (Key) | Valor (Value) | Tipo do Dado
| :--- | :---: | ---: |
| id | 1025	| Número Inteiro (Integer) |
| nome | "Maria" | Texto (String) |
| email	| "maria@email.com"	| Texto (String) |

### **Para recuperar qualquer uma dessas informações no código, você só precisa chamar a chave desejada**

O computador precisa de um formato JSON limpo. Porém, o texto vinha escrito com data = { ... } na primeira linha. Para o computador, esse data = é como um rabisco desnecessário na folha.

O código faz uma pergunta simples: "O texto começa com 'data = '?". Se sim, ele apaga esse trecho (replace) e guarda apenas o que interessa (o texto limpo do JSON).

```python
import pandas as pd
import json
from IPython.display import display

# Lendo o arquivo como texto
with open('BaseDadosJSON.json', 'r', encoding='utf-8') as f:
    conteudo = f.read().strip()

# Remove o prefixo 'data = ' se ele existir para tornar o JSON válido
if conteudo.startswith('data = '):
    conteudo_json = conteudo.replace('data = ', '', 1)
else:
    conteudo_json = conteudo

# Converte a string tratada em um dicionário e carrega no DataFrame
dados = json.loads(conteudo_json)
df = pd.DataFrame(dados)

display(df)
```

```python
import json
import pandas as pd

# 1. Solicita os novos valores ao usuário
duration = int(input("Digite o valor para 'Duration': "))
pulse = int(input("Digite o valor para 'Pulse': "))
maxpulse = int(input("Digite o valor para 'Maxpulse': "))
calories = int(input("Digite o valor para 'Calories': "))

# 2. Abre o arquivo existente e carrega os dados atuais
with open('BaseDadosJSON.json', 'r', encoding='utf-8') as f:
    conteudo = f.read().strip()

if conteudo.startswith('data = '):
    conteudo_json = conteudo.replace('data = ', '', 1)
else:
    conteudo_json = conteudo

dados = json.loads(conteudo_json)

# 3. Descobre o próximo índice numérico (chave)
proximo_indice = str(len(dados['Duration']))

# 4. Insere os novos valores no dicionário
dados['Duration'][proximo_indice] = duration
dados['Pulse'][proximo_indice] = pulse
dados['Maxpulse'][proximo_indice] = maxpulse
dados['Calories'][proximo_indice] = calories

# 5. Salva os dados atualizados de volta no arquivo mantendo o formato original 'data = ...'
with open('BaseDadosJSON.json', 'w', encoding='utf-8') as f:
    f.write('data = ' + json.dumps(dados, indent=2))

print("\nValores salvos com sucesso no arquivo 'BaseDadosJSON.json'!")

# Mostra a tabela atualizada
df_atualizado = pd.DataFrame(dados)
display(df_atualizado)
```
